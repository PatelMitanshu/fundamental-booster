```python
dict1={"name":"Alice","age":30,"city":"New York"}
dict1["name"]="Alice Smith"
#  fromkey means it is a method that creates a new dictionary with keys from an iterable and values set to a specified value. For example, if we have keys = ['a', 'b', 'c'] and value = 0, then dict.fromkeys(keys, value) will create a new dictionary {'a': 0, 'b': 0, 'c': 0}.
#  get method in dictionary is used to retrieve the value associated with a specified key. It takes the key as an argument and returns the corresponding value. If the key is not found in the dictionary, it returns a default value (which can be specified as a second argument) or None if no default value is provided. For example, dict.get('name') will return 'Alice Smith' if 'name' is a key in the dictionary, otherwise it will return None or the specified default value.
#  items method in dictionary is used to return a view object that displays a list of dictionary's key-value tuple pairs. For example, dict1.items() will return dict_items([('name', 'Alice Smith'), ('age', 30), ('city', 'New York')]).
# pop method in dictionary is used to remove a specified key and return the corresponding value. If the key is not found, it raises a KeyError unless a default value is provided. For example, dict1.pop('age') will remove the 'age' key from the dictionary and return its value (30), while dict1.pop('nonexistent_key', 'default_value') will return 'default_value' without raising an error.

```

```python
# what is membership operator in python?
# The membership operator in Python is used to test if a value is present in a sequence (such as a string, list, tuple, or set). The two membership operators are 'in' and 'not in'. The 'in' operator returns True if the value is found in the sequence, while the 'not in' operator returns True if the value is not found in the sequence. For example, 'a' in 'apple' would return True, while 'b' not in 'apple' would also return True.
```

```python
# Reverse a number
number = int(input("Enter a number: "))
reversed_number = 0
while number > 0:
    digit = number % 10
    reversed_number = reversed_number * 10 + digit
    number //= 10

reversed_number = int(str(reversed_number))
print("Reversed number:", reversed_number)
```

```python
# GCD of two numbers means the greatest common divisor of two numbers, which is the largest positive integer that divides both numbers without leaving a remainder. For example, the GCD of 12 and 15 is 3, because 3 is the largest number that can divide both 12 and 15 evenly. The GCD can be calculated using various algorithms, such as the Euclidean algorithm.
a=int(input("Enter first number: "))
b=int(input("Enter second number: "))
while b:
    a, b = b,a%b
print("GCD of the two numbers is:", a)
```

```python
# Find LCM of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = a * b
while b:
    a, b = b, a % b
lcm = c // a
print("LCM of the two numbers is:", lcm)
```
